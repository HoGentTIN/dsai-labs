## Exercise 1 - metro
At a metro station, a regular passenger records how long (in seconds) he has to wait for a train to arrive once he gets to the platform.
Find the median and the interquartile range of these times.

Help: https://numpy.org/devdocs/reference/generated/numpy.quantile.html

**Solutions**

Source: Complete Probability & Statistics 1 for Cambridge International As & A Level

# Exercise 2
The length of time, in minutes, that customers spend in a coffee shop is recorded. Calculate the following statistics:

* Median
* Range
* Quartiles
* Interquartile range  

Source: Complete Probability & Statistics 1 for Cambridge International As & A Level

```python
times = np.array([17, 15, 9, 31, 33, 41, 8, 14, 13, 22, 27, 43, 32, 14])
percentiles = [0.0,0.25,0.5,0.75,1.0]

print(f"Median", np.median(times))
print(f"Range: ", np.max(times) - np.min(times))
print(f"Quartiles", np.quantile(times, percentiles))
print(f"IQR", stats.iqr(times))
```
